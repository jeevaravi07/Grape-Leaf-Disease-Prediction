```python
%%writefile train.py
"""
Trains a grape leaf disease classifier using transfer learning (MobileNetV2).
"""

import argparse
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

def build_datasets(data_dir):
    train_ds = tf.keras.utils.image_dataset_from_directory(
        data_dir,
        validation_split=0.2,
        subset="training",
        seed=42,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
    )
    val_ds = tf.keras.utils.image_dataset_from_directory(
        data_dir,
        validation_split=0.2,
        subset="validation",
        seed=42,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
    )
    class_names = train_ds.class_names

    train_ds = train_ds.map(lambda x, y: (preprocess_input(x), y))
    val_ds = val_ds.map(lambda x, y: (preprocess_input(x), y))

    train_ds = train_ds.prefetch(tf.data.AUTOTUNE)
    val_ds = val_ds.prefetch(tf.data.AUTOTUNE)

    return train_ds, val_ds, class_names

def build_model(num_classes):
    data_augmentation = models.Sequential([
        layers.RandomFlip("horizontal"),
        layers.RandomRotation(0.1),
        layers.RandomZoom(0.1),
    ])

    base_model = MobileNetV2(
        input_shape=IMG_SIZE + (3,),
        include_top=False,
        weights="imagenet",
    )
    base_model.trainable = False

    inputs = tf.keras.Input(shape=IMG_SIZE + (3,))
    x = data_augmentation(inputs)
    x = base_model(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.3)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = tf.keras.Model(inputs, outputs)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )
    return model, base_model

def main():
    ap = argparse.ArgumentParser()
    ap.add_argument("--data_dir", default="./dataset/train")
    ap.add_argument("--epochs", type=int, default=15)
    ap.add_argument("--fine_tune_epochs", type=int, default=5)
    ap.add_argument("--out", default="grape_disease_model.keras")
    args = ap.parse_args()

    train_ds, val_ds, class_names = build_datasets(args.data_dir)
    print("Classes:", class_names)

    model, base_model = build_model(num_classes=len(class_names))
    model.summary()

    print("\n--- Phase 1: training classification head (base frozen) ---")
    model.fit(train_ds, validation_data=val_ds, epochs=args.epochs)

    print("\n--- Phase 2: fine-tuning top layers of base model ---")
    base_model.trainable = True
    for layer in base_model.layers[:-30]:
        layer.trainable = False

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )
    model.fit(train_ds, validation_data=val_ds, epochs=args.fine_tune_epochs)

    model.save(args.out)
    with open("class_names.txt", "w") as f:
        f.write("\n".join(class_names))

    print(f"\nModel saved to {args.out}")
    print("Class names saved to class_names.txt")

if __name__ == "__main__":
    main()
```

```python
%%writefile evaluate.py
"""
Evaluates a trained model on your separate test/ folder.
"""
import argparse
import tensorflow as tf
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

def main():
    ap = argparse.ArgumentParser()
    ap.add_argument("--test_dir", required=True)
    ap.add_argument("--model", default="grape_disease_model.keras")
    args = ap.parse_args()

    test_ds = tf.keras.utils.image_dataset_from_directory(
        args.test_dir,
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        shuffle=False,
    )
    class_names = test_ds.class_names
    test_ds = test_ds.map(lambda x, y: (preprocess_input(x), y))

    model = tf.keras.models.load_model(args.model)
    loss, acc = model.evaluate(test_ds)
    print(f"\nTest accuracy: {acc*100:.2f}%")
    print(f"Test loss: {loss:.4f}")
    print("Classes:", class_names)

if __name__ == "__main__":
    main()
```

```python
%%writefile predict.py
"""
Predicts the disease class of a single grape leaf image.
"""
import argparse
import numpy as np
import tensorflow as tf
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

IMG_SIZE = (224, 224)

def load_class_names(path="class_names.txt"):
    with open(path) as f:
        return [line.strip() for line in f if line.strip()]

def predict(image_path, model_path, class_names_path):
    model = tf.keras.models.load_model(model_path)
    class_names = load_class_names(class_names_path)

    img = tf.keras.utils.load_img(image_path, target_size=IMG_SIZE)
    img_array = tf.keras.utils.img_to_array(img)
    img_array = preprocess_input(img_array)
    img_array = np.expand_dims(img_array, axis=0)

    predictions = model.predict(img_array)[0]
    predicted_idx = np.argmax(predictions)
    predicted_class = class_names[predicted_idx]
    confidence = predictions[predicted_idx] * 100

    print(f"Predicted disease: {predicted_class} ({confidence:.2f}% confidence)")
    print("\nAll class probabilities:")
    for name, prob in sorted(zip(class_names, predictions), key=lambda x: -x[1]):
        print(f"  {name}: {prob*100:.2f}%")

    return predicted_class, confidence

if __name__ == "__main__":
    ap = argparse.ArgumentParser()
    ap.add_argument("--image", required=True)
    ap.add_argument("--model", default="grape_disease_model.keras")
    ap.add_argument("--class_names", default="class_names.txt")
    args = ap.parse_args()

    predict(args.image, args.model, args.class_names)
```
